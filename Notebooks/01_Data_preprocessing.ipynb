Data merging

Align with transactions

From transaction.info() output:

In transactions, the transaction ID column is id (int64, unique).

the labels use transaction_id (int64, from JSON keys).

So before merging, rename in transactions:

Clean & rename columns

Transactions

```python
transactions = transactions.rename(columns={"id": "transaction_id"})
```

```python
transactions = transactions.rename(
    columns={
        "id": "transaction_id",
        "date": "timestamp",
        "client_id": "user_id",
        "mcc": "merchant_category_code"
    }
)
```

Cards

```python
# rename
cards = cards.rename(
    columns={
        "id": "card_id",
        "client_id": "user_id"
    }
)
```

Users

```python
# rename
users = users.rename(
    columns={
        "id": "user_id"
    }
)
```

Build labels_df

```python
# Build labels DataFrame properly
# inner is the dict: {transaction_id_str: "Yes"/"No"}
inner = next(iter(labels_json.values()))

# Turn into DataFrame
labels_df = pd.DataFrame(
    list(inner.items()),
    columns=["transaction_id_str", "target_str"]
)

labels_df.head()
```

Convert to correct types

We want:

transaction_id as integer

target as 0/1 (0 = No, 1 = Yes)

```python
labels_df["transaction_id"] = labels_df["transaction_id_str"].astype("int64")
labels_df["target"] = (labels_df["target_str"] == "Yes").astype("int8")

labels_df = labels_df[["transaction_id", "target"]]

labels_df.head()
```


Merging;


```python
tx_labeled = transactions.merge(labels_df, on="transaction_id", how="left")
print(tx_labeled.shape)
tx_labeled["target"].value_counts(dropna=False)
```

```python
tx_labeled[["transaction_id", "target"]].head()
```

```python
transaction_labels_df= tx_labeled.dropna(subset=["target"]).copy()
transaction_labels_df["target"] = transaction_labels_df["target"].astype("int8")

print(transaction_labels_df.shape)

transaction_labels_df["target"].value_counts(normalize= True)
```

```python
transaction_labels__cards_df = transaction_labels_df.merge(cards, on= ["card_id", "user_id"], how= "left")
print("After cards merge:", transaction_labels__cards_df.shape)
```

Reason for ["card_id", "user_id"] as join keys: a card belongs to a single user, and this makes mismatches less likely.

```python
transaction_labels__cards_users_df = transaction_labels__cards_df.merge(users, on= "user_id", how= "left")
print("After users merge:", transaction_labels__cards_users_df.shape)
```

Merge mcc into transaction_labels__cards_users_df

```python
# convert both dtypes as object
transaction_labels__cards_users_df["merchant_category_code"] = transaction_labels__cards_users_df["merchant_category_code"].astype(str)
mcc["merchant_category_code"] = mcc["merchant_category_code"].astype(str)
```

```python
# merge
all_merged_df = transaction_labels__cards_users_df.merge(mcc, on="merchant_category_code", how="left")
print("After mcc merge:", all_merged_df.shape)
```

```python
all_merged_df.info()
```

Basic cleaning on merged table

```python
# Convert timestamp to datetime
all_merged_df["timestamp"] = pd.to_datetime(all_merged_df["timestamp"])
```

```python


#  Convert amount & numeric-like strings to float
for col in ["amount", "credit_limit", "per_capita_income", "yearly_income", "total_debt"]:
    all_merged_df[col] = (
        all_merged_df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("$", "", regex=False)  # if present
        .astype(float)
    )
```

```python
all_merged_df.isnull().sum()
```

```python
# handling missing values in merchant_state

# Fill missing merchant_state with "ONLINE" only when merchant_city is ONLINE.
# The way the dataset works: if merchant_city is 'ONLINE' then the merchant_state is blank, hence will append with merchant_state as ONLINE. 
# Fraud detection models may benefit from distinguishing between online & offline transactions.
all_merged_df.loc[all_merged_df['merchant_city'] == 'ONLINE', 'merchant_state'] = 'ONLINE'
```

```python
import numpy as np
```

```python
# Imputing missing ZIP codes using merchant-based mode imputation.
# from the data dictionary, Since ZIP codes correspond to merchants (not users), I will fill missing values
# using the most frequent ZIP per `merchant_id`. If a merchant has no known ZIPs,
# use the most common ZIP in the dataset as a fallback.

# Fill missing ZIPs with the most frequent ZIP per merchant_id
all_merged_df['zip'] = all_merged_df.groupby('merchant_id')['zip'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else np.nan))
# Fill remaining missing ZIPs with the most frequent ZIP in the dataset
all_merged_df['zip'].fillna(all_merged_df['zip'].mode()[0], inplace=True)
# Convert ZIP to integer format
all_merged_df['zip'] = all_merged_df['zip'].astype(int)
```

```python
# 3. Drop obvious columns that are not useful for modeling
drop_cols = [
    "errors",          # mostly NaN
    "card_number",     # VERY high-cardinality + leaks sensitive info
    "cvv",             # sensitive & not useful
    "address"          # high-cardinality free text
]

cols_to_drop = [c for c in drop_cols if c in all_merged_df.columns]

all_merged_df = all_merged_df.drop(columns=cols_to_drop)

all_merged_df.info()
```

```python
# check unique values in gender
all_merged_df["gender"].unique()
```

```python
all_merged_df["gender"]= all_merged_df["gender"].str.strip().str.lower()
```

```python
print(all_merged_df["gender"].unique())
```

```python
all_merged_df["use_chip"].unique()
```

```python
# Create transaction_hour first from timestamp
all_merged_df['transaction_hour'] = all_merged_df['timestamp'].dt.hour
```

```python
# create day of week for transaction from timestamp
all_merged_df['transaction_dayofweek'] = all_merged_df['timestamp'].dt.dayofweek
```

```python
# ---- Amount checks ----
non_positive_mask = all_merged_df['amount'] <= 0
amount_p99 = all_merged_df['amount'].quantile(0.99)
amount_p999 = all_merged_df['amount'].quantile(0.999)

print("===== AMOUNT CHECKS =====")
print("Non-positive amounts:", non_positive_mask.sum())
print("Amount P99:", amount_p99)
print("Amount P999:", amount_p999)
```

```python
# Non-positive amount flag
all_merged_df['amount_non_positive_flag'] = (all_merged_df['amount'] <= 0).astype(int)
```

About 449,972 transactions (~5%) have non-positive amounts, which likely correspond to refunds, reversals, or test behavior and should be kept and analyzed, not dropped. The 99th percentile is ~315.86 and the 99.9th percentile is ~890.39, so amounts above ~900 are rare and high-value, and should be treated as high-risk but informative rather than removed.

```python
age_out_of_range_mask = (all_merged_df['current_age'] < 18) | (all_merged_df['current_age'] > 100)
computed_age = all_merged_df['timestamp'].dt.year - all_merged_df['birth_year']
age_mismatch_mask = (abs(all_merged_df['current_age'] - computed_age) > 1)

print("\n===== AGE CHECKS =====")
print("Age out of range:", age_out_of_range_mask.sum())
print("Age mismatch (>1 year):", age_mismatch_mask.sum())
```

About 10,305 records have ages outside the expected 18–100 range, which is small and likely due to data entry or profile inconsistencies. However, 7.37M records (~83%) show a >1-year mismatch between current_age and birth_year. This indicates that current_age is likely not updated regularly (static field), while birth_year is reliable. For modeling, computed_age should be used as the primary age feature.

```python
#  Age out-of-range flag
all_merged_df['age_out_of_range_flag'] = (
    (all_merged_df['current_age'] < 18) | 
    (all_merged_df['current_age'] > 100)
).astype(int)

#  Age mismatch flag
all_merged_df['computed_age'] = all_merged_df['timestamp'].dt.year - all_merged_df['birth_year']
all_merged_df['age_mismatch_flag'] = (
    abs(all_merged_df['current_age'] - all_merged_df['computed_age']) > 1
).astype(int)
```

```python
# ---- Geo / location checks ----
invalid_lat_mask = ~all_merged_df['latitude'].between(-90, 90)
invalid_lon_mask = ~all_merged_df['longitude'].between(-180, 180)

invalid_zips = ['00000', '99999', '12345']
zip_str = all_merged_df['zip'].astype(str).str.zfill(5)
invalid_zip_mask = (zip_str.isin(invalid_zips)) | (zip_str.str.len() != 5)

location_suspicious_mask = invalid_zip_mask | invalid_lat_mask | invalid_lon_mask

print("\n===== LOCATION CHECKS =====")
print("Invalid latitude:", invalid_lat_mask.sum())
print("Invalid longitude:", invalid_lon_mask.sum())
print("Suspicious ZIP codes:", invalid_zip_mask.sum())
print("Location suspicious:", location_suspicious_mask.sum())
```

All latitude, longitude, and ZIP values fall within valid ranges, and no placeholder or suspicious ZIP codes were detected. Location fields appear clean and consistent, with 0 invalid entries across all checks.
